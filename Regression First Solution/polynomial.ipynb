```python
import numpy as np
import pandas as pd 
from sklearn.metrics import r2_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
dataset = pd.read_csv("Housing.csv")
dataset =dataset.drop(['guestroom','basement','hotwaterheating','airconditioning'],axis=1)
x = dataset.iloc[:,1:]
y = dataset.iloc[:,0].values
encode_column = ['mainroad','prefarea','furnishingstatus']
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(),encode_column),("standardize",StandardScaler(),[1])], remainder="passthrough")
x= ct.fit_transform(x)

x_train ,x_test, y_train,y_test=train_test_split(x,y,random_state=0, test_size=0.2)
poly = PolynomialFeatures(degree=7)
x_poly = poly.fit_transform(x_train)
model = LinearRegression()
model.fit(x_poly,y_train)
y_pred = model.predict(poly.fit_transform(x_test))
print(model.score(x_poly,y_train))
```
